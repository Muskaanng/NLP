```python
corpus="Today's date is 03-01-2026.Starting practising NLP.So, i will start by tpoic tokenisation!"
```

```python
print(corpus)
```

```python
##Tokenisation , Paragraph ---> Sentence
from nltk.tokenize import sent_tokenize
```

```python
import nltk
nltk.download('punkt_tab')
"""punkt_tab (or punkt) is a pre-trained model used for sentence boundary detection

Without this, sent_tokenize() won’t work"""
documents=sent_tokenize(corpus)
```

Sentence-wise tokenization

```python
for sentence in documents:
  print(sentence)
```

Aggressive tokenization

Splits on every punctuation

Uses regex: \w+|[^\w\s]+

```python
#paragraph --> sentences --> words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
  print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
```

```python
"""Feature	 word_tokenize	  wordpunct_tokenize
Contractions	"'s"	    ' + s
Dates	       03-01-2026	03 - 01 - 2026
Use case	   Linguistic NLP	Regex /rule-based"""
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
"""Task	                          Tokenizer
Sentence segmentation	            sent_tokenize
Classic NLP (POS, NER)	          word_tokenize
Regex-heavy text (logs)	          wordpunct_tokenize
Linguistic parsing	              TreebankWordTokenizer
Modern Transformers	              Don’t use NLTK (use model tokenizer)

Raw Text
   ↓
Sentence Tokenization
   ↓
Word Tokenization
   ↓
Stopword Removal
   ↓
Stemming / Lemmatization
   ↓
Vectorization (TF-IDF / Embeddings)
   ↓
Model
"""
```

“Tokenization converts raw text into structured units. Sentence tokenization identifies boundaries using probabilistic models like Punkt, while word tokenization applies grammar-aware rules such as Treebank to preserve linguistic meaning.”
