<h3>Stemming</h3>
Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).

What is Stemming (in plain English)

Stemming = chopping words to a base form (stem)
The stem may not be a real English word.

Goal:

Reduce different word forms to a common representation so models treat them as the same feature.

eat, eating, eaten → eat
go, goes, going → go


Used heavily in:

Text classification

Sentiment analysis

Search engines

Spam detection

Your Classification Context (Important)

Problem: Product review → Positive or Negative

Why stemming helps:

"I am eating good food"
"I eat good food"
"I have eaten good food"


➡ Model should see same meaning, not 3 different features.

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews----> eating, eat,eaten [going,gone,goes]--->go

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

<h3>PorterStemmer</h3>
“Porter Stemmer is fast and widely used but may produce non-dictionary stems.”

How it works

Rule-based

Removes common suffixes: ing, s, ed, ly, ion

Fast but aggressive

Why mistakes?

Porter does not understand meaning

It blindly applies suffix rules

history → histori is not a real word, but stemmer doesn’t care

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
  print(word + "----->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem("sitting")
```

<h3>RegexpStemmer class</h3>
NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

What this means

Remove suffixes:

ing

s

e

able

Only if word length ≥ 4

Regex matches only suffix

It doesn’t know ingeating is invalid

No linguistic intelligence

When to use RegexpStemmer

When:

You control the vocabulary

Domain-specific text (logs, tags, commands)

Avoid for:

Natural language understanding

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$',min = 4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

<h3>Snowball Stemmer</h3>
It is a stemming algorithm which is also known as the Porter2 stemming algorithm as it is a better version of the Porter Stemmer since some issues of it were fixed in this stemmer.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballsstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowballsstemmer.stem(word))
```

```python
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballsstemmer.stem("fairly"),snowballsstemmer.stem("sportingly")
```

```python
snowballsstemmer.stem('goes')
```

```python
stemming.stem('goes')
```
